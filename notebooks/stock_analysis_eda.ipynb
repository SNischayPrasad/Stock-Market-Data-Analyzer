## 1. Collect data (Yahoo Finance, falls back to CSV)

```python
raw, source = get_stock_data('AAPL', '2023-01-01', '2024-12-31')
print(source)
raw.head()
```

## 2. Clean data

```python
df, log = clean_stock_data(raw)
print(log)
df.info()
df.describe().round(2)
```

## 3. EDA + feature engineering (returns, moving averages, volatility)

```python
df = run_full_analysis(df)
df[['Close','Daily_Return','SMA_20','SMA_50','Rolling_Volatility','Drawdown']].tail(10)
```

```python
df[['Close','SMA_20','SMA_50','SMA_200']].plot(figsize=(12,5), title='Close & Moving Averages'); plt.show()
```

```python
df['Daily_Return'].hist(bins=60, figsize=(10,4)); plt.title('Daily return distribution'); plt.show()
```

## 4. Summary metrics

```python
summary = compute_summary(df, 'AAPL')
for k, v in summary.items():
    print(f'{k:<20} {v}')
```
